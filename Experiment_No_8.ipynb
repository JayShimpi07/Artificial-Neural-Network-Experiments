```python
import numpy as np
```

```python
# Define 4 bipolar vectors

pattern1 = np.array([1, 1, -1, -1])
pattern2 = np.array([-1, 1, -1, 1])
pattern3 = np.array([1, -1, 1, -1])
pattern4 = np.array([-1, -1, 1, 1])

# Store all patterns

patterns = np.array([
    pattern1,
    pattern2,
    pattern3,
    pattern4
])

print("Training Patterns:")
print(patterns)
```

```python
# Number of neurons

n = patterns.shape[1]

# Initialize weight matrix with zeros

W = np.zeros((n, n))

print("Initial Weight Matrix:")
print(W)
```

```python
# Training using Hebbian Learning Rule

for pattern in patterns:
    
    pattern = pattern.reshape(n, 1)
    
    W += np.dot(pattern, pattern.T)

# Remove self-connections

np.fill_diagonal(W, 0)

print("Final Weight Matrix:")
print(W)
```

```python
# Activation Function

def activation(x):
    
    if x >= 0:
        return 1
    else:
        return -1
```

```python
# Test input pattern (noisy input)

test_pattern = np.array([1, 1, -1, 1])

print("Test Pattern:")
print(test_pattern)
```

```python
# Recall process

output = test_pattern.copy()

for i in range(n):
    
    net_input = np.dot(W[i], output)
    
    output[i] = activation(net_input)

print("Recovered Pattern:")
print(output)
```
