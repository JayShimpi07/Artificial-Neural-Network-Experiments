```python
import numpy as np
```

```python
# Input combinations
inputs = np.array([ # All possible combinations of two binary inputs
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Weights
w1 = 1 # Weight for the first input
w2 = -1 # Weight for the second input

# Threshold
theta = 1 # Threshold for the activation function
```

```python
# McCulloch-Pitts ANDNOT Function

def mp_andnot(x1, x2):
    
    # Calculate activation value
    yin = (x1 * w1) + (x2 * w2) # Weighted sum of inputs
    
    # Apply threshold
    if yin >= theta:
        return 1
    else:
        return 0
```

```python
# Generate output for all input combinations

print("McCulloch-Pitts ANDNOT Logic")
print("x1  x2   Output")

outputs = []

for x1, x2 in inputs:
    
    y = mp_andnot(x1, x2) # Get output for the current input combination
    outputs.append(y) # Store output in a list
    
    print(x1, "  ", x2, "    ", y)
```

```python
print("Final Output:")
print(outputs)
```

```python
print("Detailed Calculation")

for x1, x2 in inputs:
    
    yin = (x1 * w1) + (x2 * w2) # Calculate activation value
    
    if yin >= theta:
        y = 1
    else:
        y = 0
    
    print(f"x1={x1}, x2={x2}, Yin={yin}, Output={y}") 
```
