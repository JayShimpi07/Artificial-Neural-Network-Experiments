```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Input values
x = np.linspace(-10, 10, 100) # Generate 100 points from -10 to 10
```

```python
# Linear Activation Function
def linear(x):
    return x # Linear activation function simply returns the input as output

# Sigmoid Activation Function
def sigmoid(x):
    return 1 / (1 + np.exp(-x)) # Sigmoid function maps input to a value between 0 and 1

# Tanh Activation Function
def tanh(x):
    return np.tanh(x) # Tanh function maps input to a value between -1 and 1

# ReLU Activation Function
def relu(x):
    return np.maximum(0, x) # ReLU function returns the input if it's positive, otherwise it returns 0

# Softmax Activation Function
def softmax(x):
    exp_x = np.exp(x - np.max(x)) # Subtracting max for numerical stability
    return exp_x / np.sum(exp_x) # Softmax function converts a vector of values to a probability distribution
```

```python
# Calculate outputs
y_linear = linear(x)
y_sigmoid = sigmoid(x)
y_tanh = tanh(x)
y_relu = relu(x)
y_softmax = softmax(x)
```

```python
plt.figure(figsize=(6,4))
plt.plot(x, y_linear)
plt.title("Linear Activation Function")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True) 
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.plot(x, y_sigmoid)
plt.title("Sigmoid Activation Function")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.plot(x, y_tanh)
plt.title("Tanh Activation Function")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.plot(x, y_relu)
plt.title("ReLU Activation Function")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.plot(x, y_softmax)
plt.title("Softmax Activation Function")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.show()
```

```python
# Example Input Array
arr = np.array([-1, 0, 1])

print("Input Array:", arr)

print("Sigmoid Output:")
print(sigmoid(arr))

print("\nTanh Output:")
print(tanh(arr))

print("\nReLU Output:")
print(relu(arr))

print("\nSoftmax Output:")
print(softmax(arr))
```
