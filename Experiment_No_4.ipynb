```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Input data

X = np.array([
    [2, 3],
    [1, 1],
    [2, 1],
    [3, 2],
    [6, 5],
    [7, 7],
    [8, 6],
    [9, 8]
])

# Target output
# -1 -> Class 1
# +1 -> Class 2

y = np.array([-1, -1, -1, -1, 1, 1, 1, 1])

print("Input Data:")
print(X)

print("\nTarget Output:")
print(y)
```

```python
# Initialize weights and bias

weights = np.zeros(2)
bias = 0

# Learning rate
eta = 0.1

# Number of iterations
epochs = 10
```

```python
# Step Activation Function

def activation_function(value):
    
    if value >= 0: # Threshold at 0
        return 1
    else:
        return -1
```

```python
# Perceptron Training

for epoch in range(epochs): # Loop over epochs
    
    for i in range(len(X)): # Loop over each training example
        
        # Calculate net input
        net_input = np.dot(X[i], weights) + bias # Calculate the net input (weighted sum + bias)
        
        # Predicted output
        y_pred = activation_function(net_input) # Apply activation function to get predicted class label
        
        # Weight update
        weights = weights + eta * (y[i] - y_pred) * X[i] # Update weights based on error and input
        
        # Bias update
        bias = bias + eta * (y[i] - y_pred) # Update bias based on error

print("Training Completed")

print("\nFinal Weights:")
print(weights)

print("\nFinal Bias:")
print(bias)
```

```python
# Plotting decision regions

plt.figure(figsize=(8,6))

# Plot class -1 points
for i in range(len(X)):
    
    if y[i] == -1:
        plt.scatter(X[i][0], X[i][1], marker='o', s=100, # Circle marker for class -1
                    label='Class -1' if i == 0 else "") 
    
    else:
        plt.scatter(X[i][0], X[i][1], marker='x', s=100, # Cross marker for class +1
                    label='Class +1' if i == 4 else "") 

# Decision boundary line

x_values = np.linspace(0, 10, 100)

# Equation of line
y_values = -(weights[0] * x_values + bias) / weights[1]

# Plot boundary line
plt.plot(x_values, y_values, linewidth=2, label='Decision Boundary')

# Graph settings
plt.xlim(0, 10)
plt.ylim(0, 10)

plt.xlabel("X1")
plt.ylabel("X2")
plt.title("Perceptron Decision Region")

plt.legend()
plt.grid(True)

plt.show()
```
